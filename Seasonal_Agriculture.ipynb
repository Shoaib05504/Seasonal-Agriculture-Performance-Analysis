### 7.1 Check Missing Values

Missing values occur when information is not available for certain records in the dataset.

Checking missing values helps identify which columns contain incomplete data and how many values are missing.

```python
missing_values = df.isnull().sum()

print("Missing Values in Each Column:\n")
print(missing_values)
```

### 7.2 Missing Value Percentage

The percentage of missing values is calculated to understand the proportion of incomplete data in each column.

This helps determine the appropriate method for handling missing values.

```python
missing_percentage = (
    df.isnull().sum() / len(df)
) * 100

print("Missing Value Percentage:\n")
print(missing_percentage)
```

### 7.3 Handling Missing Values

Missing values can affect the accuracy of data analysis and statistical calculations.

For numerical columns, missing values are replaced using the median value of the respective column.

The median is used because it is less affected by extreme values compared to the mean.

```python

df["Rainfall_mm"] = df["Rainfall_mm"].fillna(
    df["Rainfall_mm"].median()
)

df["Soil_Moisture_pct"] = df["Soil_Moisture_pct"].fillna(
    df["Soil_Moisture_pct"].median()
)

df["Yield_Tonnes_Ha"] = df["Yield_Tonnes_Ha"].fillna(
    df["Yield_Tonnes_Ha"].median()
)
for column in df.columns:
    print(column)
```

### 7.4 Verify Missing Value Handling

After handling missing values, the dataset is checked again to verify whether missing values have been successfully addressed.

```python
print("Missing Values After Handling:\n")

print(df.isnull().sum())
```

### 7.5 Check Duplicate Records

Duplicate records are identical rows that appear more than once in a dataset.

Duplicate data can affect analysis results by giving unnecessary importance to repeated observations.

# 10. Seasonal Performance Analysis

Seasonal Performance Analysis examines how agricultural performance varies across different seasons.

Seasonal conditions can influence crop yield, production, resource usage, revenue, and profit. Therefore, comparing these performance indicators across seasons can help identify the seasons that perform better or worse.

In this section, the following performance indicators are analyzed:

- Average Yield
- Average Production
- Average Revenue
- Average Profit

Different visualizations are used to compare agricultural performance across seasons and identify important seasonal patterns.

### 10.1 Average Yield by Season

Average yield is compared across different seasons to identify seasonal differences in agricultural productivity.

This analysis helps determine which season has the highest and lowest average yield.

```python
season_yield = (
    df_clean
    .groupby("season")["yield_tonnes_ha"]
    .mean()
    .sort_values(ascending=False)
)

print("Average Yield by Season:\n")
print(season_yield)
```

### Average Yield by Season Visualization

A bar chart is used to compare the average yield across different seasons.

Different colors are used for the bars to clearly distinguish the seasonal performance.

```python
plt.figure(figsize=(9, 6))

plt.bar(
    season_yield.index,
    season_yield.values,
    color=["green", "orange", "royalblue", "purple"],
    edgecolor="black"
)

plt.title("Average Agricultural Yield by Season")
plt.xlabel("Season")
plt.ylabel("Average Yield (Tonnes per Hectare)")

plt.tight_layout()

plt.show()
```

### Observation

The seasonal yield analysis shows differences in average agricultural yield across the seasons.

The season with the highest average yield is Kharif, while the season with the lowest average yield is Zaid.

This indicates that seasonal conditions may have an important influence on agricultural productivity. Further analysis of environmental factors such as rainfall, temperature, and soil moisture can help understand the reasons behind these differences.

### 10.2 Average Production by Season

Average production is analyzed across different seasons to understand how seasonal conditions affect the total quantity of agricultural output.

This comparison helps identify seasons associated with higher and lower average production.

```python
season_production = (
    df_clean
    .groupby("season")["production_tonnes"]
    .mean()
    .sort_values(ascending=False)
)

print("Average Production by Season:\n")
print(season_production)
```

### Visualization Explanation

The line chart represents the average agricultural production for each season.

- The **X-axis** represents the different agricultural seasons.
- The **Y-axis** represents the average production in tonnes.
- Each point on the line represents the average production for a particular season.
- The connecting line makes it easier to observe the variation in production between seasons.
- A higher point indicates a season with higher average production, while a lower point indicates lower average production.

```python
plt.figure(figsize=(9, 6))

plt.plot(
    season_production.index,
    season_production.values,
    marker="o",
    linewidth=2
)

plt.title("Average Production Across Seasons")
plt.xlabel("Season")
plt.ylabel("Average Production (Tonnes)")
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

### Observation

The seasonal production analysis shows clear differences in average agricultural production across the seasons.

**Kharif** has the highest average production at approximately **46.3 tonnes**, while **Zaid** has the lowest average production at approximately **38.9 tonnes**. **Rabi** has an average production of approximately **41.5 tonnes**.

This indicates that agricultural production is highest during the Kharif season and gradually decreases through Rabi and Zaid. Seasonal conditions and other factors such as rainfall, temperature, crop type, and resource availability may contribute to these differences.

### 12.9 Overall Environmental Conditions by Season

An overall environmental conditions analysis is performed by comparing rainfall, average temperature, humidity, and soil moisture across the agricultural seasons.

Since these environmental factors have different units and scales, their values are normalized before visualization.

A radar chart is used to provide an overall visual comparison of the environmental conditions across Kharif, Rabi, and Zaid seasons.

```python
environmental_summary = (
    df_clean
    .groupby("season")
    .agg({
        "rainfall_mm": "mean",
        "avg_temperature_c": "mean",
        "humidity_pct": "mean",
        "soil_moisture_pct": "mean"
    })
)

print("Average Environmental Conditions by Season:\n")
print(environmental_summary.round(2))
```

```python
environmental_normalized = (
    (environmental_summary - environmental_summary.min()) /
    (environmental_summary.max() - environmental_summary.min())
)

print("Normalized Environmental Conditions:\n")
print(environmental_normalized.round(2))
```

### Visualization Explanation

The radar chart provides an overall comparison of environmental conditions across the agricultural seasons.

- Each axis represents an environmental factor.
- The distance from the center represents the normalized value of that factor.
- Each line represents one agricultural season.
- A larger distance from the center indicates a relatively higher value for that environmental factor.
- The chart helps compare the overall environmental profiles of Kharif, Rabi, and Zaid.

```python
import numpy as np
import matplotlib.pyplot as plt

categories = environmental_normalized.columns.tolist()
N = len(categories)

angles = np.linspace(0, 2 * np.pi, N, endpoint=False).tolist()
angles += angles[:1]

fig = plt.figure(figsize=(9, 9))
ax = plt.subplot(111, polar=True)

for season in environmental_normalized.index:
    values = environmental_normalized.loc[season].tolist()
    values += values[:1]

    ax.plot(
        angles,
        values,
        linewidth=2,
        label=season
    )

    ax.fill(
        angles,
        values,
        alpha=0.1
    )

ax.set_xticks(angles[:-1])
ax.set_xticklabels(categories)

ax.set_title(
    "Overall Environmental Conditions by Season",
    pad=25
)

ax.set_ylim(0, 1)
ax.legend(loc="upper right", bbox_to_anchor=(1.25, 1.1))

plt.tight_layout()
plt.show()
```

### Observation

The overall environmental analysis shows that the three agricultural seasons have different environmental profiles.

**Kharif** shows relatively higher rainfall, humidity, and soil moisture, while **Zaid** experiences relatively higher temperature and lower soil moisture and humidity. **Rabi** generally shows moderate environmental conditions compared with the other seasons.

Overall, the analysis indicates that each season has a distinct combination of environmental conditions. These differences may influence crop selection, irrigation requirements, crop growth, and agricultural productivity.

### 14.4 Resource Usage Across Seasons

Resource consumption can vary across agricultural seasons due to differences in crop requirements, environmental conditions, and farming practices.

In this analysis, the average usage of water, fertilizer, and pesticides is compared across different seasons.

A normalized heatmap is used to compare the resource usage patterns because the resources are measured using different units and scales.

```python

pesticide_columns = [
    column for column in df_clean.columns
    if "pesticide" in column.lower()
]

print("Pesticide-related columns:")
for i, column in enumerate(pesticide_columns, start=1):
    print(f"{i}. {column}")
```

### Visualization Explanation

The heatmap shows the normalized usage of different agricultural resources across seasons.

- The **X-axis** represents the agricultural resources: Water, Fertilizer, and Pesticide.
- The **Y-axis** represents the different agricultural seasons.
- Each cell represents the normalized usage of a particular resource in a particular season.
- **Darker shades** indicate relatively higher resource usage.
- **Lighter shades** indicate relatively lower resource usage.
- Normalization allows resources with different units and scales to be compared effectively.
- The heatmap helps identify differences in resource consumption across the agricultural seasons.

```python
pesticide_columns = [
    column for column in df_clean.columns
    if "pesticide" in column.lower()
]

print("Pesticide-related columns:")
for i, column in enumerate(pesticide_columns, start=1):
    print(f"{i}. {column}")


pesticide_column = pesticide_columns[0]

season_resources = (
    df_clean
    .groupby("Season")
    .agg({
        "Water_Used_m3": "mean",
        "Fertilizer_kg_ha": "mean",
        pesticide_column: "mean"
    })
)

season_resources = season_resources.rename(
    columns={pesticide_column: "Pesticide"}
)

print("\nAverage Resource Usage by Season:\n")
print(season_resources.round(2))

resource_normalized = (
    (season_resources - season_resources.min()) /
    (season_resources.max() - season_resources.min())
)

print("\nNormalized Resource Usage:\n")
print(resource_normalized.round(2))

plt.figure(figsize=(9, 6))

plt.imshow(
    resource_normalized,
    cmap="YlOrRd",
    aspect="auto"
)

plt.colorbar(label="Normalized Usage")

plt.xticks(
    range(len(resource_normalized.columns)),
    ["Water", "Fertilizer", "Pesticide"]
)

plt.yticks(
    range(len(resource_normalized.index)),
    resource_normalized.index
)

plt.title("Normalized Resource Usage Across Seasons")
plt.xlabel("Agricultural Resources")
plt.ylabel("Season")

plt.tight_layout()
plt.show()
```

### Observation

The normalized resource usage analysis shows clear differences in resource consumption across the three agricultural seasons.

**Kharif** has the highest normalized fertilizer usage with a value of **1.00**, while its normalized water usage is **0.45** and pesticide usage is **0.38**.

**Rabi** shows relatively lower resource usage, with normalized values of **0.00 for water**, **0.32 for fertilizer**, and **0.00 for pesticide**.

**Zaid** has the highest normalized water and pesticide usage, with both values at **1.00**, while its normalized fertilizer usage is **0.00**.

Overall, **Kharif shows the highest relative fertilizer usage**, whereas **Zaid shows the highest relative water and pesticide usage**. This indicates that resource consumption patterns differ considerably across agricultural seasons.

### 16.1 Mean, Median and Standard Deviation

Mean, median, and standard deviation are important statistical measures used to understand the central tendency and variation of agricultural data.

- **Mean** represents the average value of the observations.
- **Median** represents the middle value when the observations are arranged in order.
- **Standard Deviation** measures how much the values vary from the mean.

In this analysis, these measures are calculated for important agricultural indicators such as Yield, Production, Revenue, and Profit using NumPy.

```python
yield_data = df_clean["Yield_Tonnes_Ha"].dropna().values
production_data = df_clean["Production_Tonnes"].dropna().values

revenue_column = [col for col in df_clean.columns if "revenue" in col.lower()][0]
profit_column = [col for col in df_clean.columns if "profit" in col.lower()][0]

revenue_data = df_clean[revenue_column].dropna().values
profit_data = df_clean[profit_column].dropna().values



print("Mean, Median and Standard Deviation\n")

print("Yield:")
print("Mean =", np.mean(yield_data))
print("Median =", np.median(yield_data))
print("Standard Deviation =", np.std(yield_data))

print("\nProduction:")
print("Mean =", np.mean(production_data))
print("Median =", np.median(production_data))
print("Standard Deviation =", np.std(production_data))

print("\nRevenue:")
print("Mean =", np.mean(revenue_data))
print("Median =", np.median(revenue_data))
print("Standard Deviation =", np.std(revenue_data))

print("\nProfit:")
print("Mean =", np.mean(profit_data))
print("Median =", np.median(profit_data))
print("Standard Deviation =", np.std(profit_data))
```

### Visualization Explanation

- The X-axis represents the agricultural performance indicators.
- The Y-axis represents the normalized statistical values.
- The first bar represents the Mean.
- The second bar represents the Median.
- Normalization is used because Yield, Production, Revenue, and Profit have different units and scales.
- Comparing Mean and Median helps identify differences in the distribution of each variable.
- A large difference between Mean and Median can indicate skewness or the presence of extreme observations.

```python

mean_values = np.array([
    np.mean(yield_data),
    np.mean(production_data),
    np.mean(revenue_data),
    np.mean(profit_data)
])

median_values = np.array([
    np.median(yield_data),
    np.median(production_data),
    np.median(revenue_data),
    np.median(profit_data)
])

variables = ["Yield", "Production", "Revenue", "Profit"]


combined = np.vstack([mean_values, median_values])

min_value = combined.min()
max_value = combined.max()

mean_normalized = (mean_values - min_value) / (max_value - min_value)
median_normalized = (median_values - min_value) / (max_value - min_value)

x = np.arange(len(variables))
width = 0.35

plt.figure(figsize=(10, 6))

plt.bar(x - width/2, mean_normalized, width, label="Mean")
plt.bar(x + width/2, median_normalized, width, label="Median")

plt.title("Normalized Mean vs Median")
plt.xlabel("Agricultural Performance Indicators")
plt.ylabel("Normalized Value")
plt.xticks(x, variables)
plt.legend()

plt.tight_layout()
plt.show()
```

### Observation

The Mean and Median values provide information about the central tendency of the agricultural performance indicators.

For crop Yield, the Mean is higher than the Median, indicating that some high-yield observations are increasing the average value.

The difference between the Mean and Median suggests that the Yield data is positively skewed and contains some unusually high observations.

The Standard Deviation further indicates that agricultural performance varies considerably across the observations.

Overall, Mean, Median, and Standard Deviation provide a useful statistical understanding of the central tendency and variability of the dataset.

```python
variables = ["Yield", "Production", "Revenue", "Profit"]

mean_values = np.array([
    np.mean(yield_data),
    np.mean(production_data),
    np.mean(revenue_data),
    np.mean(profit_data)
])

median_values = np.array([
    np.median(yield_data),
    np.median(production_data),
    np.median(revenue_data),
    np.median(profit_data)
])

std_values = np.array([
    np.std(yield_data),
    np.std(production_data),
    np.std(revenue_data),
    np.std(profit_data)
])

all_values = np.concatenate([
    mean_values,
    median_values,
    std_values
])

min_value = all_values.min()
max_value = all_values.max()

mean_normalized = (mean_values - min_value) / (max_value - min_value)
median_normalized = (median_values - min_value) / (max_value - min_value)
std_normalized = (std_values - min_value) / (max_value - min_value)


comparison_df = pd.DataFrame({
    "Variable": variables,
    "Mean": mean_values,
    "Median": median_values,
    "Standard Deviation": std_values
})

print("Statistical Comparison:\n")
print(comparison_df.round(2))
```

### Visualization Explanation

- The X-axis represents the agricultural indicators: Yield, Production, Revenue, and Profit.
- The Y-axis represents the normalized statistical values.
- The first bar represents the Mean.
- The second bar represents the Median.
- The third bar represents the Standard Deviation.
- Normalization allows variables with different units and scales to be compared in the same chart.
- The comparison helps identify differences between the central tendency and variability of the indicators.

```python
x = np.arange(len(variables))
width = 0.25

plt.figure(figsize=(11, 6))

plt.bar(
    x - width,
    mean_normalized,
    width,
    label="Mean"
)

plt.bar(
    x,
    median_normalized,
    width,
    label="Median"
)

plt.bar(
    x + width,
    std_normalized,
    width,
    label="Standard Deviation"
)

plt.title("Normalized Statistical Comparison")
plt.xlabel("Agricultural Indicators")
plt.ylabel("Normalized Value")
plt.xticks(x, variables)
plt.legend()

plt.tight_layout()
plt.show()
```

### Observation

The statistical comparison shows that the agricultural indicators have different levels of central tendency and variability.

The difference between the Mean and Median is particularly noticeable for Yield, indicating that the yield distribution is influenced by unusually high observations.

The Standard Deviation values show that the observations are not equally distributed around their respective means.

Overall, the comparison demonstrates that agricultural performance varies considerably across the dataset, with Yield showing a particularly uneven distribution.

### 16.6 Interquartile Range (IQR) Analysis

The Interquartile Range (IQR) is a statistical measure used to understand the spread of the middle 50% of observations.

The IQR is calculated as:

**IQR = 75th Percentile − 25th Percentile**

A larger IQR indicates greater variation among the middle 50% of observations, while a smaller IQR indicates that the observations are more closely distributed.

In this analysis, NumPy is used to calculate the 25th percentile, 50th percentile, 75th percentile, and IQR of crop yield.

```python
q1 = np.percentile(yield_data, 25)
q2 = np.percentile(yield_data, 50)
q3 = np.percentile(yield_data, 75)

iqr = q3 - q1

print("Interquartile Range Analysis of Crop Yield\n")

print("25th Percentile (Q1) =", q1)
print("50th Percentile (Median) =", q2)
print("75th Percentile (Q3) =", q3)
print("Interquartile Range (IQR) =", iqr)
```

- The difference between Q1 and Q3 represents the Interquartile Range (IQR).
- The IQR represents the spread of the middle 50% of crop yield observations.

```python
quartile_values = [q1, q2, q3]
quartile_labels = ["Q1 (25th)", "Median (50th)", "Q3 (75th)"]

plt.figure(figsize=(9, 6))

plt.plot(
    quartile_labels,
    quartile_values,
    marker="o",
    linewidth=2
)

plt.title("Quartile Analysis of Crop Yield")
plt.xlabel("Statistical Measure")
plt.ylabel("Yield (Tonnes/Ha)")

plt.tight_layout()
plt.show()
```

The **25th percentile is approximately 1.05 tonnes/ha**, while the **median is approximately 1.74 tonnes/ha** and the **75th percentile is approximately 2.83 tonnes/ha**.

Therefore, the Interquartile Range is approximately **1.78 tonnes/ha**.

The relatively large difference between the typical yield values and the previously observed maximum yield indicates that some extremely high observations exist outside the middle 50% of the data.

Overall, the IQR provides a useful measure of the typical spread of crop yield while being less affected by extreme values.

### 17.1 Correlation Matrix

A correlation matrix is used to measure the strength and direction of relationships between numerical variables.

The correlation coefficient ranges from **-1 to +1**:

- A value close to **+1** indicates a strong positive relationship.
- A value close to **-1** indicates a strong negative relationship.
- A value close to **0** indicates a weak or no linear relationship.

In this analysis, the correlation between important agricultural variables is calculated to identify their relationships.

```python
relationship_columns = [
    "Rainfall_mm",
    "Avg_Temperature_C",
    "Humidity_pct",
    "Soil_pH",
    "Soil_Moisture_pct",
    "Water_Used_m3",
    "Fertilizer_kg_ha",
    "Yield_Tonnes_Ha",
    "Production_Tonnes"
]

relationship_columns = [
    col for col in relationship_columns
    if col in df_clean.columns
]

correlation_matrix = df_clean[relationship_columns].corr()

print("Correlation Matrix:\n")
print(correlation_matrix.round(2))
```

- Negative values indicate a negative relationship.

- The color intensity helps identify stronger and weaker relationships.
- The diagonal values are 1 because each variable has a perfect correlation with itself.

```python

plt.figure(figsize=(12, 8))

plt.imshow(
    correlation_matrix,
    cmap="coolwarm",
    aspect="auto"
)

plt.colorbar(label="Correlation Coefficient")

plt.xticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns,
    rotation=45,
    ha="right"
)

plt.yticks(
    range(len(correlation_matrix.index)),
    correlation_matrix.index
)

plt.title("Correlation Matrix of Agricultural Variables")
plt.xlabel("Variables")
plt.ylabel("Variables")

plt.tight_layout()
plt.show()
```

The correlation matrix shows the strength and direction of relationships between the selected agricultural variables.

### 18.4 Overall Comparative Analysis

Overall Comparative Analysis combines the results from the seasonal, crop-wise, and state-wise comparisons.

In this analysis, the overall performance of different seasons, crops, and states is compared using four major indicators:

- Average Yield
- Average Production
- Average Revenue
- Average Profit

Since these indicators have different units and scales, the values are normalized before calculating an overall performance score.

The overall performance score helps identify the relatively better-performing season, crop, and state based on the combined performance indicators.

This provides a consolidated view of the agricultural performance observed in the dataset.

```python
performance_columns = [
    "Average_Yield",
    "Average_Production",
    "Average_Revenue",
    "Average_Profit"
]

season_normalized = (
    (seasonal_comparison - seasonal_comparison.min()) /
    (seasonal_comparison.max() - seasonal_comparison.min())
)

season_score = season_normalized.mean(axis=1)

crop_normalized = (
    (crop_comparison - crop_comparison.min()) /
    (crop_comparison.max() - crop_comparison.min())
)

crop_score = crop_normalized.mean(axis=1)

state_normalized = (
    (state_comparison - state_comparison.min()) /
    (state_comparison.max() - state_comparison.min())
)

state_score = state_normalized.mean(axis=1)

best_season = season_score.idxmax()
best_crop = crop_score.idxmax()
best_state = state_score.idxmax()

overall_comparison = pd.DataFrame({
    "Category": ["Season", "Crop", "State"],
    "Best_Performer": [
        best_season,
        best_crop,
        best_state
    ],
    "Performance_Score": [
        season_score.max(),
        crop_score.max(),
        state_score.max()
    ]
})

print("Overall Comparative Analysis:\n")
print(overall_comparison.round(2))
```

### Visualization Explanation

- The Y-axis represents the normalized overall performance score.

- The performance score is calculated using the normalized Yield, Production, Revenue, and Profit values.

- The chart provides a simple summary of the best-performing season, crop, and state.

```python
plt.figure(figsize=(10, 6))

plt.bar(
    overall_comparison["Category"],
    overall_comparison["Performance_Score"],
    color="Purple"
)

plt.title("Overall Comparative Performance")
plt.xlabel("Comparison Category")
plt.ylabel("Overall Performance Score")

plt.tight_layout()
plt.show()
```

```python
final_season_comparison = (
    df_clean
    .groupby("Season")
    .agg({
        "Yield_Tonnes_Ha": "mean",
        "Production_Tonnes": "mean"
    })
)

revenue_column = [
    col for col in df_clean.columns
    if "revenue" in col.lower()
][0]

profit_column = [
    col for col in df_clean.columns
    if "profit" in col.lower()
][0]

final_season_comparison["Average_Revenue"] = (
    df_clean.groupby("Season")[revenue_column].mean()
)

final_season_comparison["Average_Profit"] = (
    df_clean.groupby("Season")[profit_column].mean()
)


final_normalized = (
    (final_season_comparison - final_season_comparison.min()) /
    (final_season_comparison.max() - final_season_comparison.min())
)


final_normalized["Overall_Score"] = final_normalized.mean(axis=1)

print("Final Comparative Summary:\n")
print(final_season_comparison.round(2))

print("\nOverall Seasonal Performance Score:\n")
print(final_normalized["Overall_Score"].round(2))

print(
    "\nBest Overall Performing Season:",
    final_normalized["Overall_Score"].idxmax()
)

print(
    "Lowest Overall Performing Season:",
    final_normalized["Overall_Score"].idxmin()
)
```

### Visualization Explanation

- The X-axis represents the agricultural seasons.
- The Y-axis represents the overall normalized performance score.
- The score combines Yield, Production, Revenue, and Profit.
- Normalization allows indicators with different units and scales to be compared.
- A higher score indicates relatively better overall performance.
- The visualization provides a final comparison of agricultural performance across seasons.

```python
plt.figure(figsize=(10, 6))

plt.bar(
    final_normalized.index,
    final_normalized["Overall_Score"],
    edgecolor="black"
)

plt.title("Final Comparative Analysis of Seasonal Performance")
plt.xlabel("Season")
plt.ylabel("Overall Normalized Performance Score")

plt.tight_layout()
plt.show()
```

### Observation

The final comparative analysis brings together the major agricultural performance indicators.

**Kharif** shows the strongest overall performance, supported by its higher average yield, production, revenue, and profit.

**Rabi** demonstrates moderate overall performance across the major indicators.

**Zaid** shows comparatively weaker overall performance, particularly in terms of profitability.

The crop analysis showed that **Sugarcane** has very high yield and production, while **Sugarcane and Chilli** demonstrate stronger economic performance.

The geographical analysis showed noticeable differences between states, with **Punjab** performing strongly across several major indicators.

Overall, the comparative analysis indicates that agricultural performance varies considerably depending on the season, crop, and geographical location.
